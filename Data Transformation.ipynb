```python
import pandas as pd
pd.set_option("display.width", 120)
pd.set_option("display.max_columns", 30)

titanic = pd.read_csv(r"C:\Users\Daniel\Desktop\Machine learning\Titanic_Cleaned.csv")
print(titanic.shape)
titanic.head()
```

```python
df2 = titanic.copy()

# apply + lambda: number -> category
df2["fare_band"] = df2["fare"].apply(lambda x: "20%" if x >= 50 else "10%")

# map + dictionary: text -> short code
sex_map = {"male": "M", "female": "F"}
df2["sex_code"] = df2["sex"].map(sex_map)

df2[["sex", "sex_code", "fare", "fare_band"]].head()
```

```python
# assign: a new column derived from an existing one
df2 = df2.assign(fare_plus=df2["fare"] * 1.1)

# replace: relabel values
df2["embarked"] = df2["embarked"].replace("S", "Southampton")

df2[["fare", "fare_plus", "embarked"]].head()
```

```python
# rename a column
df2 = df2.rename(columns={"fare_plus": "fare_plus_10pct"})

# rename a single index label
df2.head(4).rename(index={0: "row zero"})[["sex", "fare"]]
```

## Sorting, re-indexing, ranking

```python
df2.sort_values("fare", ascending=False).head(5)
```

```python
df2.sort_values(["pclass", "fare"], ascending=[True, False]).head(6)
```

```python
df2.sort_values("age", ascending=False).reset_index(drop=True).head(3)
```

```python
df2["fare_rank"] = df2["fare"].rank(ascending=False, method="min")
df2.sort_values("fare_rank").head(5)[["fare", "fare_rank"]]
```

## Column order

```python
df2 = df2[["survived", "pclass", "sex", "sex_code", "age", "fare",
           "fare_plus_10pct", "fare_band", "fare_rank", "embarked", "alone"]]
df2.head()
```

## Grouping and aggregation

```python
df2.groupby("pclass")[["fare", "age"]].mean()
```

```python
df2.groupby("pclass")[["fare", "age"]].agg(["min", "max", "mean"])
```

```python
df2.groupby("pclass").agg(min_fare=("fare", "min"), max_age=("age", "max"))
```

```python
df2.groupby("embarked").agg({"fare": "mean", "age": "max", "survived": "mean"})
```

## Reshaping: melt / pivot

```python
wide = pd.DataFrame({"country": ["USA", "USA", "India", "India"],
                     "year": [2020, 2021, 2020, 2021],
                     "sales": [100, 120, 90, 110],
                     "profit": [20, 25, 18, 22]})
melted = wide.melt(id_vars=["country", "year"], value_vars=["sales", "profit"],
                   var_name="metric", value_name="value")
melted
```

```python
melted.pivot(index=["country", "year"], columns="metric", values="value")
```

```python
df2.groupby(["pclass", "sex"])["survived"].mean().mul(100).round(1).unstack("sex")
```

## Survival analysis

```python
new_sur = df2[df2["survived"] == 1].copy()
print(new_sur["survived"].size, "survived out of", df2["survived"].size)
print("survival rate =", round(new_sur["survived"].size / df2["survived"].size * 100, 2), "%")
```
