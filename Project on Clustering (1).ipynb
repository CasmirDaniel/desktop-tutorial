# Customer Segmentation with K-Means Clustering

## Objective
Group customers into segments based on their behavior using unsupervised
learning (K-Means). There is no target column, so the goal is to find natural
groupings in the data.

## Dataset
 File:customer_behavior.csv`
 
 Size: 200 rows, 4 columns, no missing values
 
 Features: age`, `income`, `spending_score`, `online_activity_time`

## Methodology
1. Explored the data (`info`, `describe`, null checks) and plotted a pairplot
2. Scaled all features to the 0-1 range with `MinMaxScaler`, so income (tens of
   thousands) does not dominate age or activity time
3. Fitted K-Means with **4 clusters** and added the labels as a `cluster` column
4. Visualized the clusters in a pairplot and an Income vs Spending Score scatter plot
5. Ran the elbow method (k = 1 to 9, plotting SSE/inertia) to check the choice of k

## Results
- Customers were split into 4 clusters
- SSE falls from about 66.7 (k=1) to 37.5 (k=4) and about 22.6 (k=9)

## Limitations and Future Work
- The elbow curve declines gradually with no sharp bend, so k=4 is a reasonable
  choice but not a clear-cut one
- Add a silhouette score to compare values of k
- Profile each cluster (average age, income, spending) and give each a business
  meaning, such as "high income, low spending"

```python
x = df[['age', 'income', 'spending_score', 'online_activity_time']]
```

```python
# Importing min max scaler
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
x_scale = scaler.fit_transform(x)
```

```python
# import the model for clustering
from sklearn.cluster import KMeans
```

```python
# Creating an instance of kmeans
# kmean = KMeans(n_clusters=3)
kmean = KMeans(4)
```

```python
y_pred = kmean.fit_predict(x_scale)
```

```python
y_pred = kmean.fit_predict(x_scale)
df['cluster'] = y_pred
```

```python
sns.pairplot(df, hue='cluster', palette='Set1',
             vars=['age', 'income', 'spending_score', 'online_activity_time'])
plt.show()
```

```python
df["target"] = y_pred
```

```python
df.head()
```

```python
## if you are Creating an instance of kmeans and you put it as 3 that means this process also has to be three for it to actually work.
df0 = df.loc[df["target"]==0]
df1 = df.loc[df["target"]==1]
df2 = df.loc[df["target"]==2]
df3 = df.loc[df["target"]==3]
```

```python
## Why Income vs Spending Score?
## I plotted **Income** against **Spending Score** because both were used to build the clusters, and together they show customer purchasing behavior clearly:
## **Income** shows how much a customer can afford to spend.
## **Spending Score** shows how much the customer actually spends.
plt.scatter(df0["income"], df0["spending_score"], color = "blue")
plt.scatter(df1["income"], df1["spending_score"], color = "black")
plt.scatter(df2["income"], df2["spending_score"], color = "orange")
plt.scatter(df3["income"], df3["spending_score"], color = "red")


plt.title("Clustering")
plt.ylabel("income")
plt.xlabel("spending_score")
plt.savefig('Clustering')
plt.show()
```

```python
k_rng = range(1,10)

sse = []

for i in k_rng:
    km = KMeans(i)
    km.fit_predict(x_scale)
    sse.append(km.inertia_)
sse    
```

```python
#  The full meaning of (FTBC) is Finding the Best Clusters
plt.figure(figsize = (7,6))
plt.plot(k_rng, sse)
plt.title("FTBC")
plt.savefig('FTBC')
plt.show()
```
