```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("./Customer Segmentation.csv")
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
import numpy as np
```

```python
df = df.replace('NaN', np.nan)
```

```python
df['Var_1'] = df['Var_1'].fillna(df['Var_1'].mode()[0])
df['Graduated'] = df['Graduated'].fillna(df['Graduated'].mode()[0])
df['Profession'] = df['Profession'].fillna(df['Profession'].mode()[0])
df['Family_Size'] = df['Family_Size'].fillna(df['Family_Size'].median())
df['Work_Experience'] = df['Work_Experience'].fillna(df['Work_Experience'].median())
df['Ever_Married'] = df['Ever_Married'].fillna(df['Ever_Married'].mode()[0])
```

```python
df.isnull().sum()
```

```python
# getting the distinct values of the variable.
df["Segmentation"].unique()
```

```python
df["Var_1"].unique()
```

## Label Encoding

```python
df["Spending_Score"] = df["Spending_Score"].map({"Low":0, "Average":1, "High":2})
```

```python
df["Gender"] = df["Gender"].map({"Female":0, "Male":1})
```

```python
df["Ever_Married"] = df["Ever_Married"].map({"No":0, "Yes":1})
```

```python
df["Segmentation"] = df["Segmentation"].map({"A":0, "B":1, "C":2, "D":3})
```

```python
df["Var_1"] = df["Var_1"].map({"Cat_1":0, "Cat_2":1, "Cat_3":2, "Cat_4":3,"Cat_5":4, "Cat_6":5, "Cat_7":6})
```

```python
x = df[["Age", "Gender", "Ever_Married", "Segmentation", "Var_1"]]
y = df["Spending_Score"]
```

## Optimal Feature Selection

```python
from sklearn.feature_selection import SelectKBest,f_classif
```

```python
## Instanting selectKBest
select = SelectKBest(f_classif, k = 4)
x_select = select.fit_transform(x,y)

# Getting the column names
column_name = x.columns[select.get_support()]
column_name
```

```python
x_new = df[['Age', 'Ever_Married', 'Segmentation', 'Var_1']] 
```

```python
x_transf = np.sqrt(x_new)
x_transf
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x_new, y, test_size = 0.3, random_state = 42)
xtrain.shape, xtest.shape, ytrain.shape, ytest.shape
```

## HyperParameter Tuning

```python
param = {
    'criterion':["gini", "entropy"],
    'max_depth':[3, 4, 5, 6, 7],
    'max_features':[2, 3, 4, 5, 6],
    'min_samples_split':[2,3,4]
}
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
dt = DecisionTreeClassifier(random_state = 42)
```

```python
gs = GridSearchCV(dt, param_grid = param, cv = 10)
model = gs.fit(xtrain,ytrain)
model
```

```python
y_pred = model.predict(xtest)
y_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
acc = accuracy_score(ytest,y_pred)
acc
```

```python
cm = confusion_matrix(ytest,y_pred)
cm
```
