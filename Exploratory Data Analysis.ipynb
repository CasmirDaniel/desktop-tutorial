```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("./tips.csv")
df.head()
```

```python
df.info()
```

```python
# getting the shape of the data
df.shape
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
x = df["sex"].value_counts()
```

```python
#adding title, labels to a chart using plt
plt.bar(x = x.index, height = x.values)
plt.title("Number of Male and Females")
plt.xlabel("Sex")
plt.ylabel("Number")
plt.show()
```

```python
sns.boxplot(data=df, x='size')
plt.show()
```

```python
sns.violinplot(data=df, x='size', fill=False)
plt.show()
```

```python
sns.violinplot(data=df, x='sex', y='size', inner=None)
plt.show()
```

```python
sns.swarmplot(data=df, y='tip')
plt.show()
```

```python
sns.swarmplot(data=df, x='sex', y='tip', hue='day')
plt.show()
```

```python
sns.swarmplot(data=df, x='size', y='tip', hue='sex')
plt.show()
```

```python
sns.histplot(data=df, x='tip', kde=True)
plt.show()
```

```python
sns.countplot(data=df, x='size', hue='sex')
plt.show()
```

```python
sns.countplot(data=df, x='size', hue='day')
plt.show()
```

```python
sns.boxplot(data=df, y='tip', hue='sex')
plt.show()
```

```python
sns.histplot(data=df, x='sex', color='brown')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.countplot(data=df, x='sex', hue='day')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.countplot(data=df, x='smoker', hue='day')
plt.show()
```
